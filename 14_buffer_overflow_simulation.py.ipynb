```python
while True:
    try:
        print("\n========== BUFFER OVERFLOW SIMULATION ==========")

        buffer_size = int(input("Enter buffer size: "))
        data = input("Enter data: ")

        print("\n========== OUTPUT ==========")
        print("Buffer Size :", buffer_size)
        print("Data Length :", len(data))

        if len(data) > buffer_size:
            print("Warning: Buffer limit exceeded!")
        else:
            print("Data fits safely inside the buffer.")

    except ValueError:
        print("Error: Buffer size must be a number.")
    except Exception as e:
        print("Error:", e)

    choice = input("\nC = Continue, E = Exit: ").upper()
    if choice == "E":
        break
```
