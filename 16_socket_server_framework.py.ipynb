```python
import socket

while True:
    try:
        print("\n========== SOCKET SERVER ==========")

        # Create server socket
        server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

        # Bind server to localhost and port 9000
        server.bind(("127.0.0.1", 9000))

        # Start listening
        server.listen(1)

        print("Server started successfully.")
        print("Server address : 127.0.0.1")
        print("Server port    : 9000")
        print("Waiting for client connection...")

        # Accept client connection
        client, address = server.accept()

        print("\nClient connected.")
        print("Client address :", address)

        # Receive message
        data = client.recv(1024)

        print("\n========== COMMUNICATION ==========")
        print("Client -> Server :", data.decode())

        # Send response
        response = "Message received successfully!"
        client.send(response.encode())

        print("Server -> Client :", response)

        # Close connection
        client.close()
        server.close()

        print("\nServer stopped.")

    except OSError as e:
        print("Socket Error:", e)

    except Exception as e:
        print("Error:", e)

    choice = input("\nC = Continue, E = Exit: ").upper()

    if choice == "E":
        print("Program exited.")
        break
```
