```python
while True:
    try:
        data = input("Enter text (max 10 characters): ")

        if len(data) > 10:
            print("Error: Input too long!")
        else:
            print("Accepted:", data)

    except Exception as e:
        print("Error:", e)

    if input("Continue? (y/n): ").lower() != 'y':
        print("Program exited.")
        break
```
