```python
while True:
    try:
        print("\n========== DICTIONARY OF LISTS ==========")

        students = {
            "Name": [],
            "Mark": []
        }

        # Initial student entry
        n = int(input("Enter number of students: "))

        for i in range(n):
            name = input(f"Enter name {i+1}: ")
            mark = int(input(f"Enter mark {i+1}: "))

            students["Name"].append(name)
            students["Mark"].append(mark)

        # Menu
        while True:
            print("\n========== STUDENT MENU ==========")
            print("1. Add Student")
            print("2. Update Student")
            print("3. Search Student")
            print("4. Display Students")
            print("5. Calculate Average")
            print("6. Exit")

            choice = int(input("Enter your choice: "))

            # 1. Add Student
            if choice == 1:
                name = input("Enter student name: ")
                mark = int(input("Enter student mark: "))

                students["Name"].append(name)
                students["Mark"].append(mark)

                print("Student added successfully.")

            # 2. Update Student
            elif choice == 2:
                name = input("Enter student name to update: ")

                if name in students["Name"]:
                    index = students["Name"].index(name)

                    new_mark = int(input("Enter new mark: "))
                    students["Mark"][index] = new_mark

                    print("Student mark updated successfully.")
                else:
                    print("Student not found.")

            # 3. Search Student
            elif choice == 3:
                name = input("Enter student name to search: ")

                if name in students["Name"]:
                    index = students["Name"].index(name)

                    print("Name:", students["Name"][index])
                    print("Mark:", students["Mark"][index])
                else:
                    print("Student not found.")

            # 4. Display Students
            elif choice == 4:
                print("\n========== STUDENT DETAILS ==========")

                if len(students["Name"]) == 0:
                    print("No students available.")
                else:
                    for i in range(len(students["Name"])):
                        print("Name:", students["Name"][i],
                              "| Mark:", students["Mark"][i])

            # 5. Calculate Average
            elif choice == 5:
                if len(students["Mark"]) == 0:
                    print("No marks available.")
                else:
                    average = sum(students["Mark"]) / len(students["Mark"])
                    print("Average Mark:", average)

            # 6. Exit
            elif choice == 6:
                print("Program exited.")
                break

            else:
                print("Invalid choice. Please try again.")

        break

    except ValueError:
        print("Error: Enter valid numbers.")

    except Exception as e:
        print("Error:", e)
```
