```python
from mylib import Student

s = Student()

while True:
    try:
        name = input("Enter student name: ")
        s.display(name)
    except Exception as e:
        print("Error:", e)

    if input("Continue? (y/n): ").lower() != 'y':
        break

print("Program exited.")
```
