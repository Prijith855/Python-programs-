```python
import socket

while True:
    try:
        print("\n========== HOST DETAILS ==========")

        # Enter hostname
        host = input("Enter hostname [localhost]: ")

        if host == "":
            host = "localhost"

        # Find IP address
        ip = socket.gethostbyname(host)

        # Get computer hostname
        hostname = socket.gethostname()

        # Display host details
        print("\n========== HOST DETAILS ==========")
        print("Computer Name :", hostname)
        print("Host Name     :", host)
        print("IP Address    :", ip)

    except socket.gaierror:
        print("Error: Host could not be found.")

    except Exception as e:
        print("Error:", e)

    # Search another host or exit
    choice = input("\nS = Search Another Host, E = Exit: ").upper()

    if choice == "E":
        print("Program exited.")
        break
```
