```python
count = 0

def increment_counter():
    global count
    count += 1

def decrement_counter():
    global count
    if count > 0:
        count -= 1

def reset_counter():
    global count
    count = 0


while True:
    try:
        print("\n========== COUNTER PROGRAM ==========")

        print("1. Set Counter")
        print("2. Increment Counter")
        print("3. Decrement Counter")
        print("4. Display Counter")
        print("5. Reset Counter")
        print("6. Exit")

        choice = int(input("\nEnter your choice: "))

        if choice == 1:
            count = int(input("Enter counter value: "))
            print("Counter set to:", count)

        elif choice == 2:
            increment_counter()
            print("Counter incremented.")

        elif choice == 3:
            decrement_counter()
            print("Counter decremented.")

        elif choice == 4:
            print("Current Counter:", count)

        elif choice == 5:
            reset_counter()
            print("Counter reset to 0.")

        elif choice == 6:
            print("Program exited.")
            break

        else:
            print("Invalid choice.")

    except ValueError:
        print("Error: Enter a valid number.")

    except Exception as e:
        print("Error:", e)
```
