```python
import subprocess
import platform

while True:
    try:
        print("\n========== HOST AVAILABILITY CHECK ==========")

        # Enter host address
        base = input("Enter host address [127.0.0]: ").strip()

        if base == "":
            base = "127.0.0"

        # Enter host range
        start = int(input("Enter start host number (1-254): "))
        end = int(input("Enter end host number (1-254): "))

        if start < 1 or end > 254 or start > end:
            raise ValueError("Use a valid range between 1 and 254.")

        print("\n========== OUTPUT ==========")

        # Select ping command based on operating system
        if platform.system().lower() == "windows":
            command = ["ping", "-n", "1", "-w", "1000"]
        else:
            command = ["ping", "-c", "1", "-W", "1"]

        active_hosts = []
        inactive_hosts = []

        # Check host availability
        for i in range(start, end + 1):
            ip = base + "." + str(i)

            result = subprocess.run(
                command + [ip],
                stdout=subprocess.DEVNULL,
                stderr=subprocess.DEVNULL
            )

            if result.returncode == 0:
                active_hosts.append(ip)
            else:
                inactive_hosts.append(ip)

        # Display active hosts
        print("\n========== ACTIVE HOSTS ==========")

        if active_hosts:
            for ip in active_hosts:
                print(ip, "-> Active")
        else:
            print("No active hosts found.")

        # Display inactive hosts
        print("\n========== INACTIVE HOSTS ==========")

        if inactive_hosts:
            for ip in inactive_hosts:
                print(ip, "-> Inactive")
        else:
            print("No inactive hosts found.")

    except ValueError as e:
        print("Error:", e)

    except Exception as e:
        print("Error:", e)

    choice = input("\nC = Continue, E = Exit: ").upper()

    if choice == "E":
        print("Program exited.")
        break
```
