```python
import time

while True:
    try:
        print("\n========== SYN FLOOD SIMULATION ==========")

        count = int(input("Enter number of simulated requests: "))

        if count <= 0:
            raise ValueError("Enter a positive number.")

        print("\nSimulation Output:")

        for i in range(1, count + 1):
            print("Simulated SYN request", i)
            time.sleep(0.1)

        print("\nSimulation completed.")
        print("No real network packets were sent.")

    except ValueError as e:
        print("Error:", e)
    except Exception as e:
        print("Error:", e)

    choice = input("\nC = Continue, E = Exit: ").upper()
    if choice == "E":
        break
```
