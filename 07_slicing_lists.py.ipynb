```python
while True:
    try:
        print("\n========== LIST SLICING ==========")

        values = input("Enter elements separated by space: ")
        numbers = [int(x) for x in values.split()]

        start = int(input("Enter start index: "))
        end = int(input("Enter end index: "))

        print("\n========== OUTPUT ==========")

        # Display the list
        print("Original List :", numbers)

        # Display first element
        print("First Element :", numbers[0])

        # Display last element
        print("Last Element  :", numbers[-1])

        # Slice using start and end
        print("Sliced List   :", numbers[start:end])

        # Display alternate elements
        print("Alternate     :", numbers[::2])

        # Reverse the list
        print("Reverse List  :", numbers[::-1])

    except ValueError:
        print("Error: Enter valid numbers and indexes.")

    except IndexError:
        print("Error: The list is empty.")

    except Exception as e:
        print("Error:", e)

    choice = input("\nC = Continue, E = Exit: ").upper()

    if choice == "E":
        print("Program exited.")
        break
```
