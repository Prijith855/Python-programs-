```python
import socket

while True:
    try:
        print("\n========== BANNER GRABBING ==========")

        # Enter authorized host
        host = input("Enter authorized host [127.0.0.1]: ").strip()

        if host == "":
            host = "127.0.0.1"

        # Enter port number
        port = int(input("Enter port number: ").strip())

        # Connect to service
        s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        s.settimeout(3)

        print("\nConnecting to service...")

        s.connect((host, port))

        # Send HTTP request
        s.sendall(b"HEAD / HTTP/1.0\r\nHost: localhost\r\n\r\n")

        # Retrieve service banner
        banner = s.recv(1024)

        print("\n========== SERVICE BANNER ==========")

        if banner:
            print(banner.decode(errors="ignore"))
        else:
            print("No banner received from the service.")

        s.close()

    except ValueError:
        print("Error: Enter a valid port number.")

    except ConnectionRefusedError:
        print("Error: No service is running on this port.")

    except socket.timeout:
        print("Error: Connection timed out.")

    except socket.gaierror:
        print("Error: Host could not be found.")

    except Exception as e:
        print("Error:", e)

    choice = input("\nC = Continue, E = Exit: ").upper()

    if choice == "E":
        print("Program exited.")
        break
```
