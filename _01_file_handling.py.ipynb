```python
import socket

while True:
    try:
        print("\n========== UDP SERVER AND CLIENT ==========")

        server = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        server.bind(("127.0.0.1", 9999))

        print("UDP Server started...")
        print("Server address: 127.0.0.1:9999")

        client = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)

        message = input("\nEnter message from client: ")
        client.sendto(message.encode(), ("127.0.0.1", 9999))

        data, client_address = server.recvfrom(1024)

        print("\n========== SERVER ==========")
        print("Message received :", data.decode())
        print("Client address   :", client_address)

        response = "Message received successfully!"
        server.sendto(response.encode(), client_address)

        response_data, server_address = client.recvfrom(1024)

        print("\n========== COMMUNICATION ==========")
        print("Client  -> Server :", data.decode())
        print("Server  -> Client :", response_data.decode())
        print("Server address    :", server_address)
        print("Client address    :", client_address)

        client.close()
        server.close()

    except OSError as e:
        print("Socket Error:", e)

    except Exception as e:
        print("Error:", e)

    choice = input("\nC = Continue, E = Exit: ").upper()

    if choice == "E":
        print("Program exited.")
        break
```
