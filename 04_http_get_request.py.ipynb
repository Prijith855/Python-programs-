```python
import urllib.request
import ssl

while True:
    try:
        print("\n========== HTTP GET REQUEST ==========")

        url = input("Enter URL [https://example.com]: ")

        if url == "":
            url = "https://example.com"

        # Disable SSL certificate verification
        context = ssl._create_unverified_context()

        # Send GET request
        response = urllib.request.urlopen(url, context=context, timeout=5)

        # Read response content
        data = response.read(200).decode(errors="ignore")

        print("\n========== OUTPUT ==========")
        print("Status Code :", response.status)
        print("URL         :", response.geturl())

        print("\n========== RESPONSE HEADERS ==========")
        print(response.headers)

        print("\n========== RESPONSE CONTENT ==========")
        print(data)

    except urllib.error.URLError as e:
        print("HTTP Error:", e)

    except Exception as e:
        print("Error:", e)

    choice = input("\nC = Continue, E = Exit: ").upper()

    if choice == "E":
        print("Program exited.")
        break
```
